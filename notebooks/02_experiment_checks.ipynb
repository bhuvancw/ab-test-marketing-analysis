```python
import pandas as pd
from scipy.stats import chi2_contingency

df = pd.read_csv('../data/processed/ab_test_clean.csv')
df.shape
```

```python
sizes = df['test_group'].value_counts()
print(sizes)
print((sizes / len(df)).round(3))
```

```python
df.groupby("test_group")["total_ads"].describe()
```

```python
pd.crosstab(df["most_ads_day"], df["test_group"], normalize="columns").round(3)
```

```python
pd.crosstab(df["most_ads_hour"], df["test_group"], normalize="columns").round(3)
```

```python
table = pd.crosstab(df["most_ads_day"], df["test_group"])
chi2, p, dof, expected = chi2_contingency(table)
print("p-value:", p)
```

```python
df["total_ads"].quantile([0.5, 0.9, 0.99, 0.999])
```

```python
print("users with more than 500 ads:", (df["total_ads"] > 500).sum())
print(df[df["total_ads"] > 500].groupby("test_group")["converted"].mean())
```

```python
# does conversion differ by day, and does it differ by group within a day?
by_day = df.groupby(["most_ads_day", "test_group"])["converted"].mean().unstack()
by_day.round(4)
```

```python
hour_table = pd.crosstab(df["most_ads_hour"], df["test_group"])
chi2, p, dof, expected = chi2_contingency(hour_table)
print("p-value:", p)
```
