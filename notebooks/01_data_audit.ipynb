##### 01 - Data Audit 
Goal: check the quality and structure of the marketing A/B test data before any analysis.

```python
import pandas as pd

df = pd.read_csv('../data/raw/marketing_AB.csv')

print(df.shape)
df.head()
```

```python
df.columns
```

```python
df = df.drop(columns=['Unnamed: 0'])
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.info()
```

```python
print(df.isnull().sum())
print('duplicate user ids: ', df['user_id'].duplicated().sum())
print(df['test_group'].value_counts())
print(df['test_group'].value_counts(normalize=True))
print(df.groupby('test_group')['converted'].sum())
```

```python
df['total_ads'].describe()
```

```python
print(df['most_ads_day'].value_counts())
print(df['most_ads_hour'].value_counts().sort_index())
```

```python
df['converted'].value_counts()
```

```python
summary = df.groupby("test_group")["converted"].agg(["count", "sum", "mean"])
summary.columns = ["users", "conversions", "conv_rate"]
summary
```

```python
df.to_csv("../data/processed/ab_test_clean.csv", index=False)
```
